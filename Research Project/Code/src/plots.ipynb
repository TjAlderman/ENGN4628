```python
from state import HEV
import matplotlib.pyplot as plt
import numpy as np
```

```python
v = HEV()
P_ic = np.linspace(0,v.max_ic_power)
ice_e = v.efficiency(P=P_ic,motor="ICE")
P_ev = np.linspace(0,v.max_ev_power)
ev_e = v.efficiency(P=P_ev,motor="EV")

plt.figure(dpi=300)
plt.plot(P_ic,1/ice_e,label='IC Cost')
plt.plot(P_ev,1/ev_e,label='EV Cost')
plt.yscale('log')
plt.legend()
plt.title("Cost versus Power")
plt.ylabel("Cost")
plt.xlabel('Power (kW)')
plt.tight_layout()
plt.show()
```
